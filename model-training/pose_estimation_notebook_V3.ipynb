```python
import scipy.io as sio
import os

base = r'C:\Ahmed\UNI\SEM 8\FYP\FYP--2\Dataset\mpi_inf_3dhp\dataset\mpi'

total_frames = 0
for s in range(1, 9):
    for seq in range(1, 3):
        mat_path = os.path.join(base, f'S{s}', f'Seq{seq}', 'annot.mat')
        mat = sio.loadmat(mat_path)
        frames = mat['annot3'][0, 0].shape[0]
        print(f"S{s}/Seq{seq}: {frames:,} frames")
        total_frames += frames

print(f"\nTotal frames: {total_frames:,}")
print(f"With 4 cameras each: {total_frames:,} training samples")
```
